```python
import csv
from tensorboard.backend.event_processing import event_accumulator

ea = event_accumulator.EventAccumulator('/home/hzchen/jyh/train_vanilla_clip_Qwen2.5-Math-7B-16k-think_openr1/training/events.out.tfevents.1769009596.gnode02.2582932.0')
ea.Reload()

# 导出所有标量数据到 CSV
with open('/home/hzchen/jyh/verl/training_data.csv', 'w', newline='') as f:
    writer = csv.writer(f)
    writer.writerow(['tag', 'step', 'value', 'wall_time'])
    
    for tag in ea.Tags().get('scalars', []):
        for event in ea.Scalars(tag):
            writer.writerow([tag, event.step, event.value, event.wall_time])

print("已导出到 training_data.csv")
```
